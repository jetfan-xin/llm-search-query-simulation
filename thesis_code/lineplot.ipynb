```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
# 引用了seaborn后，图片有了背景，但会报找不到字符，中文无法显示,需手动添加
sns.set_theme(style="whitegrid")
# 两种添加字符均可
sns.set(font="SimHei")

data_str_1 = '''
[0.0907118439651399, 0.12747570548912981, 0.4555137965482791]
[0.07800140095594654, 0.09781376451830982, 0.31539686463928907]
[0.06996861360861387, 0.08861728826728857, 0.407860606060606]
[0.05982686757686779, 0.07420766439442927, 0.36467099567099553]
'''
data_str_2 = '''
[0.08485926807839562, 0.12025040790293898, 0.4407855927133155]
[0.0749539947070669, 0.10337895905646252, 0.3805450871361278]
[0.06943701825318725, 0.09358308751598751, 0.4349473495328857]
[0.05705228455754839, 0.07905068193885775, 0.40116747541544184]
'''
data_str_3 = '''
[0.5772920591461247, 0.6147678360240213, 0.8044359673713816]
[0.5764631529799615, 0.607949908213182, 0.7641062077247736]
[0.5809110555648803, 0.6070079032659531, 0.7902774229049683]
[0.5686414869725704, 0.5896915210485458, 0.7579162615537643]
'''

# 解析字符串数据
data_list_1 = [[float(num) for num in line.strip('[]\n').split(', ')] for line in data_str_1.strip().split('\n')]
data_list_2 = [[float(num) for num in line.strip('[]\n').split(', ')] for line in data_str_2.strip().split('\n')]
data_list_3 = [[float(num) for num in line.strip('[]\n').split(', ')] for line in data_str_3.strip().split('\n')]

# 修改以下代码，请改为水平绘制data_list_1，data_list_2，data_list_3的3张折线图，指标分别为Jaccard 相似度、BLEU和BERTScore
# 创建 DataFrame
df_1 = pd.DataFrame(data_list_1, columns=["1 随机选择", "2 频率选择", "3 标准"])
df_2 = pd.DataFrame(data_list_2, columns=["1 随机选择", "2 频率选择", "3 标准"])
df_3 = pd.DataFrame(data_list_3, columns=["1 随机选择", "2 频率选择", "3 标准"])
# 绘制折线图
plt.figure(figsize=(14, 6))
# 第一个图
plt.subplot(1, 3, 1)
sns.lineplot(data=df_1, dashes=False, palette="Set1", legend=False)
# plt.xlabel("查询次数")
plt.ylabel("Jaccard 相似度")
plt.ylim(0, 0.5)
plt.xticks(range(len(df_1)), ['第1次查询', '第2次查询', '第3次查询', '>=4次查询'])

# 第二个图
plt.subplot(1, 3, 2)
sns.lineplot(data=df_2, dashes=False, palette="Set1", legend=False)
# plt.xlabel("查询次数")
plt.ylabel("BLEU")
plt.ylim(0, 0.5)
plt.xticks(range(len(df_2)), ['第1次查询', '第2次查询', '第3次查询', '>=4次查询'])

# 第三个图
plt.subplot(1, 3, 3)
sns.lineplot(data=df_3, dashes=False, palette="Set1")
# plt.xlabel("查询次数")
plt.ylabel("BERTScore")
plt.ylim(0.5, 1)
plt.xticks(range(len(df_3)), ['第1次查询', '第2次查询', '第3次查询', '>=4次查询'])
plt.legend(loc='upper right')  # 添加图注
plt.tight_layout()
plt.show()
```
