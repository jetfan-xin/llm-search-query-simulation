```python
%matplotlib inline
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme(style="whitegrid")
# 两种添加字符均可

plt.set(font="SimHei", font_scale=1.4)
```

```python
def draw_boxplot(design):
    f_1 = f"./evaluation/new/{design}_jaccard.csv"
    data = pd.read_csv(f_1, encoding="gb18030")#防止弹出警告
    df_1 = pd.DataFrame(data)

    f_2 = f"./evaluation/new/{design}_bleu.csv"
    data = pd.read_csv(f_2, encoding="gb18030")#防止弹出警告
    df_2 = pd.DataFrame(data)

    f_3 = f"./evaluation/new/{design}_bert.csv"
    data = pd.read_csv(f_3, encoding="gb18030")#防止弹出警告
    df_3 = pd.DataFrame(data)

    plt.figure(figsize=(20, 6))

    plt.subplot(1, 3, 1)
    sns.boxplot(x='Jaccard相似度', y='name', hue='position', data=df_1, palette="Set1")
    plt.xlim(0, 1)  # 设置x轴的范围为0到1
    sns.despine(offset=10, trim=True)
    plt.ylabel("")
    plt.legend([],[], frameon=False)

    plt.subplot(1, 3, 2)
    sns.boxplot(x='BLEU', y='name', hue='position', data=df_2, palette="Set1")
    plt.xlim(0, 1)  # 设置x轴的范围为0到1
    sns.despine(offset=10, trim=True)
    plt.ylabel("")
    plt.legend(loc='upper right')
    plt.yticks([])

    plt.subplot(1, 3, 3)
    sns.boxplot(x='BERTScore', y='name', hue='position', data=df_3, palette="Set1")
    sns.despine(offset=10, trim=True)
    plt.ylabel("")
    plt.yticks([])
    plt.legend([],[], frameon=False)
    # 设置图例和标题
    plt.tight_layout()
    plt.show()
```

```python
draw_boxplot("actions")
```

```python
draw_boxplot("prompt")
```
