```python
def draw_boxplot(data, data_name, metric_name):
    # 准备数据
    values = list(data.values())
    data["all"] = [value for sublist in values for value in sublist]
    task_labels = list(data.keys())
    all_values = list(data.values())

    # 绘制箱线图
    plt.figure(figsize=(10, 6))
    plt.boxplot(all_values, labels=task_labels)
    plt.title(f'{metric_name}')
    plt.xlabel('任务')
    # plt.ylabel(f'{metric_name}')
    plt.grid(True)
    plt.show()
```

```python
import json
import matplotlib.pyplot as plt

def plot_boxplot(data):
    tasks = list(data.keys())
    scores = list(data.values())

    plt.figure(figsize=(10, 6))
    plt.boxplot(scores, labels=tasks, showfliers=True, flierprops=dict(marker='o', markersize=8, markerfacecolor='red', linestyle='none'))
    plt.title('Evaluation Metric Scores')
    plt.xlabel('Tasks')
    plt.ylabel('Score')
    plt.grid(True)
    plt.show()

# 读取JSON数据文件
with open('data.json', 'r') as file:
    json_data = json.load(file)

# 绘制箱线图
plot_boxplot(json_data)
```

```python
import os

def count_txt_files(root_dir):
    total_txt_files = 0

    # 遍历root_dir及其子文件夹
    for root, dirs, files in os.walk(root_dir):
        # 对每个文件进行检查
        for file in files:
            # 检查文件是否以'.txt'为后缀
            if file.endswith('.txt'):
                total_txt_files += 1

    return total_txt_files

# 指定文件夹路径
folder_path = './data/OCRtxtzhongnew'

# 统计txt文件个数
total_txt_files = count_txt_files(folder_path)
print(f'Total number of txt files: {total_txt_files}')
```

```python
from bert_score import BERTScorer
import torch

# 定义模拟查询和真实查询列表
simulated_queries = ["可持续生活方式", "国际知名的动物保护组织", "动物保护协会"]
real_queries = ["真实查询1", "可持续生活方式", "国际知名的动物保护组织"]

# 实例化BERTScorer，指定使用的模型和语言
scorer = BERTScorer(model_type="bert-base-chinese", lang="zh", batch_size=1, device='cuda' if torch.cuda.is_available() else 'cpu')

# 计算BERTScore
for sim_query, real_query in zip(simulated_queries, real_queries):
    P, R, F1 = scorer.score([sim_query], [real_query])
    print(f"模拟查询: {sim_query}, 真实查询: {real_query}, F1得分: {F1.item()}")
```
