```python
# The user has provided the data from the image in text form.
# We can now parse this data and create the requested multi-level bar chart.

# First, let's create a DataFrame from the provided data.
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
plt.rcParams['font.sans-serif'] = 'SimHei'
```

```python
# I apologize for the confusion. Let's correct the bar chart with the proper labeling and grouping as specified.
# This time, we'll group by the first-level title "User Background" and then plot each category within that background.

# First, let's create a dictionary structure for the data
data = {
    '性别': {'女': 2.6, '男': 2.2},
    '学院专业': {
        '经济/商/金融': 3.1,
        '人工智能': 1.9,
        '统计': 2.9,
        '文史哲': 2.1,
        '信息学': 2.0,
        '其他': 2.4
    },

    '对任务主题的了解程度': {'1': 2.4, '2': 2.7, '3': 2.1, '4': 2.7, '5': 2.5},
    '对任务的感兴趣程度': {'1': 2.2, '2': 2.7, '3': 2.7, '4': 2.4, '5': 2.0},
    '预计完成任务困难程度': {'1': 2.3, '2': 2.5, '3': 2.7, '4': 2.2, '5': 2.2}
}
```

```python
# Define the positions for the group of bars
positions = {}
current_position = 0
for background, categories in data.items():
    positions[background] = list(range(current_position, current_position + len(categories)))
    current_position += len(categories) + 1  # We leave a space between groups

# Now let's create the bar chart
fig, ax = plt.subplots(figsize=(12, 5))

# Initialize an offset for x position, and prepare lists to store xticks and labels
offset = 0
xticks = []
xticklabels = []

# Plot each set of values with an increasing offset, and collect xticks and labels
for background, sub_categories in data.items():
    x_values = np.arange(len(sub_categories)) + offset
    ax.bar(x_values, list(sub_categories.values()), label=background)
    xticks.extend(x_values)
    xticklabels.extend(sub_categories.keys())

    # Update the offset, adding 1 for spacing between main categories
    offset += len(sub_categories) + 1

# Set the labels and title
ax.set_ylabel('一项任务上的平均查询次数', fontsize=14)

# Apply the xticks and labels
ax.set_xticks(xticks)
ax.set_xticklabels(xticklabels, rotation=45, ha='right', fontsize=14)

# Add legend outside the plot to avoid overlap
ax.legend(fontsize=12,bbox_to_anchor=(1,1))

plt.tight_layout()
plt.show()
```
